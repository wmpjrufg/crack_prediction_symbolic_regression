# Dataset statistics

```python
df_x = df.describe().T
df_x.drop(columns=["count", "25%", "50%", "75%"], inplace=True)
df_x
```

```python
df_x_latex = df_x.copy()
df_x_latex.columns = [str(c).replace("_", " ") for c in df_x_latex.columns]
df_x_latex.index = [str(i).replace("_", " ") for i in df_x_latex.index]

latex_table = df_x_latex.to_latex(
    index=True,
    float_format="%.2f",
    caption="Descriptive statistics of Dataset 1.",
    label="tab:dataset_1_stats_summary",
    longtable=False,
    escape=False
)

print(latex_table)
```
