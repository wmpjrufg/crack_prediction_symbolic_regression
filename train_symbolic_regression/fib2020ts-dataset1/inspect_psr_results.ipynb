# Inspect PySR Results - Complete

```python
import pickle
import pandas as pd
from sympy import sympify, latex

filename = 'results_pysr_model_8020_fib2020ts_dataset_1_rv3.pkl'
with open(filename, 'rb') as f:
    results = pickle.load(f)

rows = []
for i, r in enumerate(results[1:], 1):
    m = r['model']
    rows.append({'index': i, 'train_rmse': m['train_rmse'], 'test_rmse': m['test_rmse'],
                 'train_r2': m['train_r2'], 'test_r2': m['test_r2'], 'eq': m['best_equation']})

df = pd.DataFrame(rows).sort_values('test_r2', ascending=False)
print(f'Models: {len(df)}')
print('\nStatistics:')
print(df[['train_rmse', 'test_rmse', 'train_r2', 'test_r2']].describe())
```

## Top 10 models

```python
print('\nTop 10:')
print(df.head(10)[['index', 'train_r2', 'test_r2', 'train_rmse', 'test_rmse']].to_string())

df.head(10)[['index', 'train_rmse', 'test_rmse', 'train_r2', 'test_r2', 'eq']].to_csv('best_10_models.csv', index=False)
print('\nExported: best_10_models.csv')
```

## Variable mapping - Geometric

```python
geometric = {
    'diametro_da_armadura_mm': 'phi_l',
    'resistencia_do_aco_fy_mpa': 'f_y',
    'numero_de_barras_em_uma_camada': 'n_b',
    'espacamento_entre_as_barras_mm': 's_b',
    'resistencia_a_tracao_do_concreto_mpa': 'f_t',
    'cobrimento_mm': 'cob',
    'distancia_ao_centro_de_gravidade_da_armadura_mm': 'd_p',
    'largura_da_viga_b_mm': 'b_w',
    'altura_da_viga_h_mm': 'h',
    'area_de_aco_tracionada_as_mm2': 'A_s',
    'momento_fletor_ms_knm': 'M_s',
    'taxa_de_armadura_conforme_mc2020': 'rho_s',
    'tensao_na_armadura_menos_tension_stiffening_variavel_mpa': 'sigma_sts',
}

print('Geometric variables:')
for k, v in sorted(geometric.items()):
    print(f'  {k:50} -> {v}')
```

## Variable mapping - Crack width models (w_*)

```python
crack_models = {
                    'fib_model_code_2010': 'w_fib',
                    'eurocode_2': 'w_ec',
                    'nbr_6118_wk1_fissuracao_estabilizada': 'w_nbr1',
                    'nbr_6118_wk2_fissuracao_nao_estabilizada': 'w_nbr2',
                    'clark_1956': 'w_clark',
                    'desayiganesan_1985': 'w_dg',
                    'oh_kang_1987': 'w_ok',
                    'caldas_1997': 'w_caldas',
                    'gergely_e_lutz_1968': 'w_gl',
                    'equacao_15_do_artigo_de_belarbi_e_hsu_1994': 'w_bh',
                    'beta_conforme_fields_e_bischoff_2004': 'w_beta',
                    'slip_de_yuan_et_al_2025': 'w_yuan',
                    'slip_de_biscaia_e_soares_2020': 'w_bs',
                }

print('\nCrack width models:')
for k, v in sorted(crack_models.items()):
    print(f'  {k:50} -> {v}')

all_map = {**geometric, **crack_models}
print(f'\nTotal: {len(all_map)} variables')
```

## Clean LaTeX function

```python
def clean_latex_overleaf(s):
    # Remove operatorname{square} and operatorname{cube}
    s = s.replace(r'\operatorname{square}\left(', '(')
    s = s.replace(r'\operatorname{cube}\left(', '(')
    return s
```

## Convert top 3 to LaTeX

```python
out = {}
for i, r in df.head(2).iterrows():
    eq = r['eq'].replace('y = ', '')
    for k, v in all_map.items():
        eq = eq.replace(k, v)
    try:
        lat = latex(sympify(eq))
        lat = clean_latex_overleaf(lat)
        out[len(out)+1] = {'r2': r['test_r2'], 'rmse': r['test_rmse'], 'latex': lat}
    except Exception as e:
        print(f'Error: {e}')

print('\n' + '='*80)
print('TOP 3 - OVERLEAF READY')
print('='*80)

for k, v in out.items():
    print(f'\nRank {k}: R2={v["r2"]:.6f}')
    print(f'\\begin{{equation}}')
    print(f'w = {v["latex"]}')
    print(f'\\end{{equation}}')
```

## Export files

```python
with open('equations_latex_overleaf.txt', 'w', encoding='utf-8') as f:
    f.write('TOP 3 EQUATIONS - OVERLEAF READY\n')
    f.write('='*80 + '\n\n')
    for k, v in out.items():
        f.write(f'Rank {k}: R2={v["r2"]:.6f}\n')
        f.write(f'\\begin{{equation}}\n')
        f.write(f'w = {v["latex"]}\n')
        f.write(f'\\end{{equation}}\n\n')

print('Exported: equations_latex_overleaf.txt')
```

## Export selected model

Index of the discrepancy model adopted in the paper. The JSON is read by the notebooks in `test_model_error`.

```python
import json

selected_index = 4
r = results[selected_index]
m = r['model']
best = m['model_object'].get_best()
eq = m['best_equation']
eq_symbols = eq
for k, v in all_map.items():
    eq_symbols = eq_symbols.replace(k, v)
selected = {
                'index': selected_index,
                'equation': eq,
                'latex': clean_latex_overleaf(latex(sympify(eq_symbols))),
                'complexity': int(best['complexity']),
                'score': float(best['score']),
                'train_r2': m['train_r2'],
                'test_r2': m['test_r2'],
                'train_rmse': m['train_rmse'],
                'test_rmse': m['test_rmse'],
                'scaler': {k: v for k, v in r.items() if isinstance(v, dict) and 'avg' in v},
            }
with open('selected_model.json', 'w', encoding='utf-8') as f:
    json.dump(selected, f, indent=4)

print(f"Index: {selected['index']} | complexity: {selected['complexity']} | score: {selected['score']:.4f}")
print(f"R2 train: {selected['train_r2']:.3f} | R2 test: {selected['test_r2']:.3f}")
print(f"w = {selected['latex']}")
print('Exported: selected_model.json')
```
