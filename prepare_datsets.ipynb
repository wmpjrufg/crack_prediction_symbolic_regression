# Libraries

```python
import pickle
import pandas as pd
pd.set_option('display.max_columns', None)
import re
import unicodedata
```

# Read dataset

```python
df = pd.read_excel('data_models_rv2.xlsx', sheet_name='Dataset_2')
def clean_column_name(name):
    name = unicodedata.normalize('NFKD', str(name)).encode('ASCII', 'ignore').decode('ASCII')
    name = name.lower()
    name = name.replace(' ', '_')
    name = re.sub(r'[^\w_]', '', name)
    name = re.sub(r'__+', '_', name)
    name = name.strip('_')
    return name
df.columns = [clean_column_name(col) for col in df.columns]
```

# Clean dataset

```python
df.drop(['autor', 'output_w_exp_mm', 'fib_model_code_2020_mm', 'fib_model_code_2020_com_ts_variavel_mm', 'wexp_menos_wfib2020ts'], inplace=True, axis=1)
df.head()
```

```python
x = df.drop('wexp_menos_wfib2020', axis=1)
x.head()
```

```python
y = df['wexp_menos_wfib2020'].to_frame()
y.head()
```

# Pickle data

```python
data_pickle = [{'metadados': 'index [1] contain independent variables (x), index [2] contain dependent variables (y)'}, x, y]
```

```python
filename = f'dataset_2.pkl'
with open(filename, 'wb') as f:
    pickle.dump(data_pickle, f)
```
