```python
import pandas as pd

# # 1. Instalar dependências
# !pip install scikit-learn openpyxl matplotlib seaborn pandas

# 2. Importar bibliotecas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler

# Modelos de regressão
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
df = pd.read_excel("/content/m.xlsx")
# df = df[df["wexp (mm)"] <= 0.4]
df
```

```python
# 5. Escolher coluna alvo (Y)
target_column = "w (mm)"  # 🔁 Altere aqui se quiser prever outra variável

# 6. Separar features e target
X = df.drop(columns=[target_column])
y = df[target_column]

# 7. Preprocessamento: remover colunas categóricas (se houver) e padronizar
X = pd.get_dummies(X, drop_first=True)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 8. Separar em treino/teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 9. Lista de modelos
models = {
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(),
    "Lasso": Lasso(),
    "RandomForest": RandomForestRegressor(),
    "GradientBoosting": GradientBoostingRegressor(),
    "KNN": KNeighborsRegressor()
}

# 10. Treinar e avaliar modelos
results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    results.append({"Model": name, "R2": r2})

# 11. Mostrar tabela com R²
results_df = pd.DataFrame(results).sort_values(by="R2", ascending=False)
print("R² dos modelos:")
print(results_df)

# 12. Exportar em LaTeX
latex_code = results_df.to_latex(index=False, float_format="%.3f")
with open("r2_models_table.tex", "w") as f:
    f.write(latex_code)

# 13. Gráfico de valores preditos vs observados para o melhor modelo
best_model_name = results_df.iloc[0]['Model']
best_model = models[best_model_name]
y_pred_best = best_model.predict(X_test)

plt.figure(figsize=(8,6))
sns.scatterplot(x=y_test, y=y_pred_best)

# ✅ Adiciona linha vermelha tracejada na diagonal
min_val = min(y_test.min(), y_pred_best.min())
max_val = max(y_test.max(), y_pred_best.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', label='Ideal Prediction')

plt.xlabel("Observed")
plt.ylabel("Predicted")
plt.title(f"Observed vs Predicted - {best_model_name}")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score

# 1. Carregar e limpar os dados
df = pd.read_excel("m.xlsx")  # Substitua pelo caminho local se necessário
df.dropna(inplace=True)
# df = df[df["w (mm)"] <= 0.4]
print(len(df))

# 2. Definir variável alvo
target_column = "w (mm)"
X = df.drop(columns=[target_column])
y = df[target_column]

# 3. Preprocessamento: dummies e normalização
X = pd.get_dummies(X, drop_first=True)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Separar em treino/teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2)

# 5. Criar e treinar modelo Gradient Boosting
model = GradientBoostingRegressor()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# 7. CoV dos resíduos (erros)
residuals = np.abs(y_pred/y_test)
std_residual = np.std(residuals)
mean_residual = np.mean(residuals)

# 8. Métrica de desempenho
r2 = r2_score(y_test, y_pred)

# 9. Resultados
print(f"🎯 R² do Gradient Boosting: {r2:.3f}")
print(f"🎯 StD dos resíduos: {std_residual:.3f}")
print(f"🎯 Média dos resíduos: {mean_residual:.3f}")
```
