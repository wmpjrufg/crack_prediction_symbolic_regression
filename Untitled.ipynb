```python
import pandas as pd


df = pd.read_excel("./datset.xlsx")

df.head()
```

```python
from sklearn.model_selection import train_test_split

df = df.drop("Autor", axis=1)

X = df.drop("w_exp (mm)", axis=1)
y = df["w_exp (mm)"]

X = X.rename(
    columns=lambda c: c.encode("ascii", "ignore").decode("ascii")  # remove accents
)
X.columns = [
    c.replace("(", "").replace(")", "").replace(" ", "_") for c in X.columns
]


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=33)
```

```python
from sklearn.model_selection import GridSearchCV
from pysr import PySRRegressor
import numpy as np

# Define the base model
model = PySRRegressor(
    binary_operators=["+", "-", "*", "/"],
    unary_operators=["exp", "sin", "cos", "log", "sqrt"],   # pick only operators that make sense for your domain
    progress=True,
)

# Define search grid
search_params_grid = {
    # Evolution budget
    "niterations": [100, 300, 500],
    "population_size": [200, 500, 1000],

    # Complexity control
    "maxsize": [15, 30, 50],

    # Genetic operators
    "crossover_probability": [0.3, 0.5, 0.7],

    # Parsimony pressure
    "parsimony": [0.0, 0.01, 0.05],
}

# Wrap with GridSearchCV
grid = GridSearchCV(
    estimator=model,
    param_grid=search_params_grid,
    scoring="neg_root_mean_squared_error",  # or "r2"
    cv=3,                                   # 3-fold cross-validation
    n_jobs=-1,                              # parallelize CV across cores
    verbose=2,
)

# Fit search
grid.fit(X_train, y_train)

print("✅ Best Params:", grid.best_params_)
print("📉 Best Score:", grid.best_score_)

model = grid.best_model_
```

```python
from sklearn.metrics import r2_score, mean_squared_error as MSE

y_pred_test =  model.predict(X_test)
y_pred_train = model.predict(X_train)

rmse_test = MSE(y_pred_test, y_test)**(1/2)
rmse_train = MSE(y_pred_train, y_train)**(1/2)


r2_test = r2_score(y_test, y_pred_test)
r2_train = r2_score(y_train, y_pred_train)

print("Root Mean Squared Error train", rmse_train)
print("Root mean Squared Error Test", rmse_test)

print("R^2 Score train", r2_train)
print("R^2 Score test", r2_test)
```
