```python
import pandas as pd


df = pd.read_excel("./datset.xlsx")

df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = df.drop("Autor", axis=1)

X = df.drop("w_exp (mm)", axis=1)
y = df["w_exp (mm)"]

X = X.rename(
    columns=lambda c: c.encode("ascii", "ignore").decode("ascii")
)
X.columns = [
    c.replace("(", "").replace(")", "").replace(" ", "_") for c in X.columns
]


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=33)
```

```python
from pysr import PySRRegressor

model = PySRRegressor(
    parsimony=0.02,
    maxsize=20,
    maxdepth=6,
    population_size=20,
    ncycles_per_iteration=3100,
    adaptive_parsimony_scaling=100.0,
    
    deterministic=False, 
    parallelism='multithreading', 
    
    progress=True,
    
    niterations=1200,
    should_optimize_constants=True,
    binary_operators=["+", "-", "*", "/", "pow"],
    unary_operators=["sin", "cos", "exp", "log", "sqrt", "square"],
    timeout_in_seconds=600
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

y_pred_test = model.predict(X_test)
y_pred_train = model.predict(X_train)

rmse_train = (mean_squared_error(y_train, y_pred_train))**(1/2)
rmse_test = (mean_squared_error(y_test, y_pred_test))**(1/2)

r2_test = r2_score(y_pred_test, y_test) 
r2_train = r2_score(y_pred_train, y_train)

print("RMSE Train: ", rmse_train)
print("RMSE test: ", rmse_test)

print("R2 test: ", r2_test)
print("R2 train: ", r2_train)
```

```python
import matplotlib.pyplot as plt

train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.scatter(y_train, train_pred, alpha=0.6)
plt.plot([y_train.min(), y_train.max()], [y_train.min(), y_train.max()], 'r--')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Training Set')

plt.subplot(1, 2, 2)
plt.scatter(y_test, test_pred, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Test Set')

plt.tight_layout()
plt.show()
```

```python
best_eq = model.get_best()
print(f"Best equation: {best_eq['equation']}")
print(f"Complexity: {best_eq['complexity']} nodes")

print("\nFull equations found:")
print(model.equations_[['complexity', 'loss', 'score', 'equation']])
```
